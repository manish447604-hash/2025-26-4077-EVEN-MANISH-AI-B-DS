```python
# LIST
fruits = ["apple", "banana", "orange"]

print(fruits)
print(fruits[0])   # first item

fruits.append("mango")
print(fruits)


# TUPLE
colors = ("red", "green", "blue")

print(colors)
print(colors[1])   # second item

for color in colors:
    print(color)


# SET
numbers = {1, 2, 3, 4}

print(numbers)

numbers.add(5)
numbers.remove(2)

if 3 in numbers:
    print("3 is in the set")

for num in numbers:
    print(num)


# DICTIONARY
student = {
    "name": "Alice",
    "age": 20,
    "course": "Python"
}

print(student)
print(student["name"])

student["grade"] = "A"   # ✅ Fixed
student["age"] = 21

for key, value in student.items():
    print(key, ":", value)
```
